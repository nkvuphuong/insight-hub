## 1. Setup

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

REPORT_NAME = 'report-name'  # kebab-case, dùng cho tên file export
```

## 2. Load Data

SQL đã verified qua MCP. Yêu cầu Claude export kết quả vào `queries/result.csv` trước khi chạy cell này.

```python
df = pd.read_csv('queries/result.csv')

print(f'Rows: {len(df)}, Cols: {list(df.columns)}')
df.head()
```

## 3. Plotly Chart (Interactive POC)

Dùng df trực tiếp — không transform.

```python
fig = px.line(
    df,
    x='month',         # tên column từ SQL
    y='revenue',       # tên column từ SQL
    title='[Report Name]',
    labels={'month': 'Month', 'revenue': 'Revenue'},
)
fig.update_layout(hovermode='x unified')
fig.show()
```

## 4. Matplotlib Export (Static PNG)

```python
fig_static, ax = plt.subplots(figsize=(12, 6))

ax.plot(df['month'], df['revenue'], marker='o')
ax.set_title('[Report Name]', fontsize=14, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Revenue')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x:,.0f}'))
plt.xticks(rotation=45)
plt.tight_layout()

out = f'{REPORT_NAME}.png'
fig_static.savefig(out, dpi=150)
print(f'Saved: {out}')
plt.show()
```
